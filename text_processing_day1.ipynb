```python
def solution(n, edges, user):
    graph = []


    for _ in range(n+1):
        graph.append([])

    for a, b in edges:
        graph[a].append(b)
        graph[b].append(a)


    friends = set(graph[user])

    recommendations = set()

    for friend in graph[user]:
        for candidate in graph[friend]:
            if candidate != user and candidate not in friends:
                recommendations.add(candidate)




    return sorted(list(recommendations))

print(solution(
    6,
    [[1, 2], [1, 3], [2, 4], [3, 4], [3, 5], [5, 6]],
    1
))
```

### TF 문서 안의 단어 빈도
###  TF는 Term Frequency의 줄임말, 특정 문서안에서 특정 단어가 얼마나 자주 나왔는 지를 나타냄

### TF(t,d )는 t는 단어, d는 문서

### 괄호 안의 기호는 어떤 단어를 어느 문서에서 볼 것인가를 지정한다.
### 같은 단어라도 문서에 따라 TF가 달라질 수 있다.

### 예를들어 : TF( t, d) = 문서 d에서 단어가 t가 등장한 횟수를 의미한다면
### A : " 베송 지연 배송에 대한 불만"
### B: 배송이 나타난 위치 두곳을 찾음
### 2. TF(배송, A) = 2
### 3. 지연과 불만은 각각 1회이므로 TF는 1
### 4.  이 문서에 없는 판손은 TF = 0


### DF 단어가 등장한 문서 수
### DF 는 Dcoument Frequency 의 줄임말, 특정 단어가 한번 이상 등장한 문서의 개수
### 문서 마다 해당 단어가 있다, 없다를 확인하고 있다면 확인 한 문서 수를 센다.
### DF(t) 는 단어 t에 대한 문서 수를 의미한다.

### 배송 : 문서 1에 있음 1 + 문서 2에 있음 1 + 문서 3에 있음 1 = DF 3
### 파손 : 문서 1에 없음 0 + 문서 2에 없음 0 + 문서 3에 있음 1 = DF 1


### TF와 구분 하는 별도 예시를 보면
###  A: " 배송, 배송, 배송" 이고 B는 " 배송 지연" 이면
### TF(배송, A) =3 이지만 DF(배송)= 2이다.


### IDF의 기본 역문서 빈도
### IDF는 Inverse Document Frequnecy의 줄임말
### 한 단어가 전체 문서에서 얼마나 드물게 나타났는 지를 반영하는 값

### 동일한 문서 모음에서 DF가 적을 수록 IDF는 커진다.

### 계산식을 보면
### 배송 : 3/ 3 = 1 ln(1)= 0
### 지연 3 / 2 = 1.5 ln(1.5) = 0.405
### 파손 3/1 = 3 -> ln(3) = 1.099
### 같은 N에서 파손의 Df가 가장 작으므로 IDF가 가장 큰값
### 모든 문서에 등장하는 배송은 기본 공식에서는 0 ( 공식에 따라 0 이 아닐 수 있음)
### 0 이라고 중요하지 않다는 것이 아니라 문서들 사이에서 문서를 구분하는 정보가 작다라고 해석하는 것이 적절하다.

### 상품이라는 키워드가 여러 리뷰에 공통적으로 있지만 파손이라는 키워드가 일부 리뷰에만 있다면
### 파손이 해당 리뷰들을 구분하는 단서가 될 수 있다는 의미다.

### TF-IDF 의 계산
### TF-IDF 는 특정 분석의 단어빈도 TF와 전체 문서에서의 희소성을 반영하는 IDF를 곱한 값

### 문서 안에서 자주 쓰이는 주요 단어를 파악하고, 다른 문서에서는 상대적으로 드문 단어의 특징적인 키워드를 찾는데 활용 된다.


### 계산식 
### TF-IDF (t, d) x IDF(t)
### 문서 3 " 배송 지연 파손" 에서는 세 단어의 TF가 각각 1
### 배송 : 1 x 0 = 0 / 지연 1 x ln(1.5) = 0.405 / ln(3) = 1.099

### 가중치는 계산에서 어떤 요소를 어느 정도로 반영할 지를 나타내는 수치

```python
import numpy as np

query_vector = np.array([1.0, 0.0])

document_vectors = np.array([
    [0.8, 0.6],
    [0.6, 0.8],
    [0.0, 1.0]
])

# 1. 코사인 유사도 계산
dot_products = document_vectors @ query_vector

query_norm = np.linalg.norm(query_vector)

document_norms = np.linalg.norm(
    document_vectors,
    axis=1
)

similarities = (
    dot_products
    / (query_norm * document_norms)
)



```

```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# 1. 검색 대상 문서
data = pd.DataFrame([
    {
        "doc_id": "Q01",
        "text": "Failure is a lesson, not the end.",
        "tags": "learning"
    },
    {
        "doc_id": "Q02",
        "text": "Every setback is a chance to try again.",
        "tags": "retry"
    },
    {
        "doc_id": "Q03",
        "text": "Small steps can create great change.",
        "tags": "progress"
    },
    {
        "doc_id": "Q04",
        "text": "Friendship is built on kindness.",
        "tags": "friendship"
    },
    {
        "doc_id": "Q05",
        "text": "Failure means you should stop trying.",
        "tags": "giving-up"
    }
])

# 2. 임베딩 모델 준비
model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

# 3. 검색 대상 문서 임베딩 생성
document_vectors = model.encode(
    data["text"].tolist()
)

print("문서 임베딩 행렬 크기:")
print(document_vectors.shape)

# 4. 사용자 질문 임베딩 생성
query = "How can I keep going after repeated failures?"

query_vector = model.encode([query])

# 5. 코사인 유사도 계산
similarities = cosine_similarity(
    query_vector,
    document_vectors
)[0]

# 6. 문서별 유사도 저장
data["similarity"] = similarities

# 7. Top-3 유사 문서 검색
top_k = 3

results = (
    data.sort_values(
        "similarity",
        ascending=False
    )
    .head(top_k)
)

print("\n사용자 질문:")
print(query)

print("\nTop-3 검색 결과:")
print(
    results[
        ["doc_id", "text", "tags", "similarity"]
    ].round(3).to_string(index=False)
)
```

```python
from openai import OpenAI
from dotenv import load_dotenv


# 1. 시스템 프롬프트
system_prompt = """
당신은 명언을 추천하는 AI입니다.

제공된 참고 문서만 사용하여 답변하세요.
질문에 적합한 명언 하나를 추천하세요.
명언의 문서 ID와 추천 이유를 함께 작성하세요.
"""

# 2. 검색 결과라고 가정한 가상의 문서
documents = [
    {
        "id": "Q01",
        "text": "Small steps can lead to progress."
    },
    {
        "id": "Q02",
        "text": "Mistakes are opportunities to learn and try again."
    },
    {
        "id": "Q03",
        "text": "Kindness helps build strong relationships."
    }
]

# 3. 참고 문맥 구성
context = "\n".join(
    f"{doc['id']}: {doc['text']}"
    for doc in documents
)

# 4. 사용자 질문
user_question = """
실패했을 때 다시 도전하는 데
도움이 되는 명언을 추천해 줘.
"""

# 5. 사용자 프롬프트 구성
user_prompt = f"""
[참고 문서]
{context}

[사용자 질문]
{user_question}
"""

# 6. API에 전달할 메시지 구성
messages = [
    {
        "role": "system",
        "content": system_prompt
    },
    {
        "role": "user",
        "content": user_prompt
    }
]

# 7. 실제 전달할 메시지 내용 확인
print("=== API 입력 메시지 ===")

for message in messages:
    print(f"\n[{message['role']}]")
    print(message["content"].strip())

load_dotenv()

client = OpenAI()

response = client.responses.create(
    model="gpt-5.6-luna",
    input=messages
)

print("\n=== LLM 응답 ===")
print(response.output_text)
```
